```python
import unicodedata
import pandas as pd
 
 
# ==============================
# 🔹 1. FUNCION PARA LIMPIAR TEXTO
# ==============================
def limpiar(texto):
    if pd.isna(texto):
        return ""
    texto = str(texto).lower()
    texto = unicodedata.normalize('NFKD', texto).encode('ascii', 'ignore').decode('utf-8')
    return texto
 
# 🔹 Categorías de ejemplo (reemplaza por tus propias categorías y palabras clave)
categorias_columna_principal = {
    "Categoria A": ["palabra clave a1"],
    "Categoria B": ["palabra clave b1"],
    "Categoria C": ["palabra clave c1"],
    "Categoria D": ["palabra clave d1"],
    "Categoria E": ["palabra clave e1"],
    "Categoria F": ["palabra clave f1"],
    "Categoria G": ["palabra clave g1"],
    "Categoria H": ["palabra clave h1"],
    "Categoria I": ["palabra clave i1"],
    "Categoria J": ["palabra clave j1"],
    "Categoria K": ["palabra clave k1"],
    "Categoria L": ["palabra clave l1"],
    "Categoria M": ["palabra clave m1"],
    "Categoria N": ["palabra clave n1"],
    "Categoria O": ["palabra clave o1"],
    "Categoria P": ["palabra clave p1"]
}
 
categorias_columna_secundaria = {
    "Categoria Especial 1": ["palabra clave especial"]
}
 
# 🔹 Función clave
def clasificar(fila):
    resultado = set()
 
    # 🔸 Evaluar columna principal (ej: entidad o campo categórico primario)
    texto_principal = limpiar(fila["Columna_Principal"])  # 👈 cambia si tu columna se llama distinto
    partes = texto_principal.split(",")
 
    for parte in partes:
        for categoria, palabras in categorias_columna_principal.items():
            for palabra in palabras:
                if palabra in parte:
                    resultado.add(categoria)
                    break
 
    # 🔸 Evaluar columna secundaria (ej: tipo o clasificación especial)
    texto_secundario = limpiar(fila["Columna_Secundaria"])  # 👈 cambia por el nombre real
    for categoria, palabras in categorias_columna_secundaria.items():
        for palabra in palabras:
            if palabra in texto_secundario:
                resultado.add(categoria)
                break
 
    return ", ".join(sorted(resultado))
 
# 🔹 Aplicación
df = pd.read_excel(r"ruta_archivo.xlsx", sheet_name="Hoja1")
 
df["Categorias"] = df.apply(clasificar, axis=1)
 
df.to_excel("resultado.xlsx", index=False)
 
print("✅ Proceso terminado. Archivo generado: resultado.xlsx")
```
